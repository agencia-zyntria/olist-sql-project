## Fase 1: comprobaciones finales

Antes de limpiar miramos tres cosas: los pagos raros (`not_defined` y valor 0), el pedido que no tiene pago y si las tablas "encajan" entre sí (integridad referencial).

```python
# Estado de cada pedido, para saber de qué tipo son los pagos raros
estado = orders.set_index("order_id")["order_status"]

raros = payments[(payments["payment_type"] == "not_defined") | (payments["payment_value"] == 0)].copy()
raros["order_status"] = raros["order_id"].map(estado)
display(raros)

# ¿Qué pedido no tiene pago? (~ significa NO)
sin_pago = orders[~orders["order_id"].isin(payments["order_id"])]
display(sin_pago[["order_id", "order_status"]])
```

```python
# Integridad referencial: ¿todo lo que apunta a otra tabla existe en ella?
print("Pedidos con cliente inexistente:", (~orders["customer_id"].isin(customers["customer_id"])).sum())
print("Reseñas de pedidos inexistentes:", (~reviews["order_id"].isin(orders["order_id"])).sum())
print("Pagos de pedidos inexistentes:", (~payments["order_id"].isin(orders["order_id"])).sum())

# Duplicados exactos (filas idénticas)
print("Filas idénticas -> customers:", customers.duplicated().sum(),
      "| reviews:", reviews.duplicated().sum(),
      "| payments:", payments.duplicated().sum())
```

### Qué nos dicen las comprobaciones

- **Pagos `not_defined` (3):** todos valen 0 y son de pedidos **cancelados** → no son pagos reales.
- **Pagos de valor 0 (9):** 3 son esos `not_defined`; los otros 6 son **vales (voucher)** de pedidos normales → son reales, se quedan.
- **Pedido sin pago (1):** es un pedido entregado. Se queda en `orders` (no afecta a nuestras preguntas).
- **Integridad referencial:** todo es 0 → podemos poner las **claves foráneas** en MySQL sin errores.
- **Duplicados exactos:** ninguno en ninguna tabla.

## Fase 2: limpieza

### Paso 1: convertir fechas de texto a fecha

Las fechas venían como texto (`str`), y con texto no se pueden restar fechas para calcular retrasos. `pd.to_datetime` las convierte. Los huecos pasan a ser `NaT` ("Not a Time"), que en MySQL será `NULL`.

```python
fechas_orders = ["order_purchase_timestamp", "order_approved_at",
                 "order_delivered_carrier_date", "order_delivered_customer_date",
                 "order_estimated_delivery_date"]
for col in fechas_orders:
    orders[col] = pd.to_datetime(orders[col])

for col in ["review_creation_date", "review_answer_timestamp"]:
    reviews[col] = pd.to_datetime(reviews[col])

print(orders[fechas_orders].dtypes)
print("Primera compra:", orders["order_purchase_timestamp"].min())
print("Última compra:", orders["order_purchase_timestamp"].max())
```

### Paso 2: código postal como texto

pandas leyó `customer_zip_code_prefix` como número y le quitó los ceros a la izquierda (`01151` → `1151`). Un código postal es una **etiqueta**, no una cantidad, así que lo pasamos a texto y rellenamos con ceros hasta 5 caracteres (`zfill(5)`).

```python
print("Códigos con menos de 5 dígitos:", (customers["customer_zip_code_prefix"].astype(str).str.len() < 5).sum())

customers["customer_zip_code_prefix"] = customers["customer_zip_code_prefix"].astype(str).str.zfill(5)

display(customers.head())
```

### Paso 3: una sola reseña por pedido

**547 pedidos** tienen más de una reseña (551 filas de más), y en **202** de ellos las notas son distintas. Si los dejamos, esos pedidos cuentan doble en la nota media (pregunta 2).

**Decisión:** nos quedamos con la reseña **más reciente** (`review_answer_timestamp` más alto), porque es la opinión final del cliente.

Después de esto `order_id` es único en reviews → puede ser la **clave primaria** de la tabla en MySQL. (`review_id` no sirve: 814 se repiten porque una misma reseña puede estar asignada a varios pedidos).

```python
reviews_limpio = (reviews
    .sort_values("review_answer_timestamp")           # de más antigua a más reciente
    .drop_duplicates(subset="order_id", keep="last")  # por pedido, la última
)

print("Antes:", len(reviews), "| Después:", len(reviews_limpio))
print("Pedidos distintos:", reviews_limpio["order_id"].nunique())
print("¿order_id ahora es único?:", reviews_limpio["order_id"].is_unique)
```

### Paso 4: pagos — quitar los `not_defined`

Los 3 pagos `not_defined` valen 0 y son de pedidos **cancelados**: no son un pago real, así que los quitamos.

Los otros 6 pagos de valor 0 son **vales (voucher)** de pedidos normales: un vale de 0 es raro pero real, así que **se quedan**.

Clave primaria de payments: no hay una sola columna única, pero la pareja **(order_id, payment_sequential)** sí lo es → será una **clave compuesta** en MySQL.

```python
print("pagos not_defined:", (payments["payment_type"] == "not_defined").sum())

payments_limpio = payments[payments["payment_type"] != "not_defined"].copy()

print("Antes:", len(payments), "| Después:", len(payments_limpio))
print("¿(order_id, payment_sequential) único?:", not payments_limpio.duplicated(["order_id", "payment_sequential"]).any())
```

### Paso 5: guardar las tablas limpias en `datos/limpios/`

`index=False` evita que pandas guarde una columna extra con el número de fila. Estos CSV son los que cargaremos en MySQL con `to_sql`.

```python
limpios = Path("../datos/limpios")
limpios.mkdir(exist_ok=True)

orders.to_csv(limpios / "orders.csv", index=False)
customers.to_csv(limpios / "customers.csv", index=False)
reviews_limpio.to_csv(limpios / "order_reviews.csv", index=False)
payments_limpio.to_csv(limpios / "order_payments.csv", index=False)

for f in sorted(limpios.glob("*.csv")):
    print("guardado:", f.name)
```
