```python
import pandas as pd
from sqlalchemy import create_engine
from getpass import getpass

# 1. Configura tu conexión (la contraseña se pide al ejecutar, no se guarda en el código)
usuario = "root"
password = getpass("Contraseña de MySQL: ")
host = "localhost"
puerto = "3306"
base_datos = "olist"

# Creamos el túnel de conexión con utf8mb4 para acentos en portugués
engine = create_engine(
    f"mysql+pymysql://{usuario}:{password}@{host}:{puerto}/{base_datos}?charset=utf8mb4"
)

# 2. Lista de tus tablas EN EL ORDEN OBLIGATORIO por las claves foráneas
ruta = "../datos/limpios/"

# Primero: CUSTOMERS (forzando el código postal como texto de 5 dígitos)
print("Cargando customers...")
df_customers = pd.read_csv(f"{ruta}customers.csv", dtype={"customer_zip_code_prefix": str})
df_customers.to_sql("customers", con=engine, if_exists="append", index=False, chunksize=5000)

# Segundo: ORDERS
print("Cargando orders...")
df_orders = pd.read_csv(f"{ruta}orders.csv")
df_orders.to_sql("orders", con=engine, if_exists="append", index=False, chunksize=5000)

# Tercero: ORDER_REVIEWS
print("Cargando order_reviews...")
df_reviews = pd.read_csv(f"{ruta}order_reviews.csv")
df_reviews.to_sql("order_reviews", con=engine, if_exists="append", index=False, chunksize=5000)

# Cuarto: ORDER_PAYMENTS
print("Cargando order_payments...")
df_payments = pd.read_csv(f"{ruta}order_payments.csv")
df_payments.to_sql("order_payments", con=engine, if_exists="append", index=False, chunksize=5000)

print("¡Las 4 tablas se han cargado correctamente!")
```
